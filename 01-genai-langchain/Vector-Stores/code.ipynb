```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_chroma import Chroma
```

```python
from langchain_core.documents import Document
```

```python
doc1=Document(
    page_content="Virat kohli is an indian criketer and a batsman",
    metadata={"team":"Rcb"}
)
doc2=Document(
    page_content="ms dhoni is an indian criketer and a wikket keeper and batsman",
    metadata={"team":"csk"}
)
doc3=Document(
    page_content="jasprit bumrah is an indian criketer and a bowler",
    metadata={"team":"Rcb"}
)

docs=[doc1,doc2,doc3]
```

```python
embeddings = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001",
    google_api_key="your_actual_api_key_here"
)
```

```python
vector_store = Chroma(
    collection_name="sample",
    embedding_function=embeddings,
    persist_directory="chroma_db",
)
```

```python
##add documents
vector_store.add_documents(docs)
```

```python
## View documents
vector_store.get(include=['embeddings','documents'])
```

```python
## Search document
vector_store.similarity_search(
    query="who are the batsmen in indian cricket team?",
    k=2
)
```

```python
vector_store.similarity_search_with_score(
    query="who are the batsmen in indian cricket team?",
    k=2
)
```

```python
## Update document
update_doc1=Document(
    page_content="Virat kohli is an indian criketer and a batsman and he was a former captain of rcb",
    metadata={"team":"Rcb"}
)
vector_store.update_document(document_id='8225bb78-5676-458e-ac64-889119e20010',document=update_doc1)
```

```python
## delete document
vector_store.delete(ids=['8225bb78-5676-458e-ac64-889119e20010'])
vector_store.get(include=['embeddings','documents'])
```
